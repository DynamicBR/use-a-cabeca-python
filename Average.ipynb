```python
FN = "Darius-13-100m-Fly.txt"
```

```python
FOLDER = "swimdata/"
```

```python
with open(FOLDER + FN) as file:
    lines = file.readlines()
```

```python
print(dir(lines))
```

```python
times = lines[0].strip("").split(",")
print(times)
```

```python
#Code Book
coverts = []
for t in times:
    minutes, rest = t.split(":")
    seconds, hundredths = rest.split(".")

    print(t, "->", (int(minutes) * 60 * 100) + (int(seconds) * 100) + int(hundredths))

    converted_time = (int(minutes) * 60 * 100) + (int(seconds) * 100) + int(hundredths)

    coverts.append(converted_time)

print(coverts)
```

```python
import statistics

average = statistics.mean(coverts)

min_secs, hundredths = str(round(average / 100, 2)).split(".")
min_secs = int(min_secs)

minutes = min_secs // 60

seconds = min_secs - minutes*60

average = str(minutes) + ":" + str(seconds) + "." + hundredths

print(average)
```
